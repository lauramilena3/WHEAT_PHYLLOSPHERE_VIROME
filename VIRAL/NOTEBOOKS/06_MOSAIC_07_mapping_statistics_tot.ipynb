# Mapping Statistics

```python
# Read data from qc_read_counts into a DataFrame
df_counts_paired = pd.read_csv(qc_read_counts, index_col=0)

# Select columns "sample" and "bbduk" from df_counts_paired
reads_df = df_counts_paired[["sample", "bbduk"]]

# Create a new column "bbduk2" in reads_df, using a lambda function to calculate the minimum value between 2000000 and each value in "bbduk"
reads_df['bbduk2'] = reads_df['bbduk'].apply(lambda x: min(2000000, x))

# Rename columns in reads_df
reads_df.columns = ["sample", "read_count", "bbduk2"]

# Initialize empty lists
mapped_pair_f = []
mapped_pair_u = []
mapped_pair_v = []
mapped_pair_a = []
mapped_pair_p = []

# Iterate over SAMPLES
for sample in SAMPLES:
    # Read content from files into corresponding lists
    content_f = open(mapping_dir + "/bowtie2_flagstats_filtered_" + sample + "." + sampling + '.txt').readlines()
    content_u = open(mapping_dir + "/STATS_FILES/bowtie2_flagstats_filtered_" + sample + "_unfiltered_contigs." + sampling + '.txt').readlines()
    content_v = open(mapping_dir + "/STATS_FILES/bowtie2_flagstats_filtered_" + sample + "_viral_contigs." + sampling + '.txt').readlines()
    content_a = open(mapping_dir + "/STATS_FILES/bowtie2_flagstats_filtered_" + sample + "_assembled_contigs." + sampling + '.txt').readlines()
    content_p = open(mapping_dir + "/REFERENCES/bowtie2_flagstats_filtered_active_vOTUs_threshold_1p5_" + sample + "." + sampling + '.txt').readlines()

    # Append calculated values to the respective lists
    mapped_pair_f.append(int(content_f[1].split()[0]) / 2)
    mapped_pair_u.append(int(content_u[1].split()[0]) / 2)
    mapped_pair_v.append(int(content_v[1].split()[0]) / 2)
    mapped_pair_a.append(int(content_a[1].split()[0]) / 2)
    mapped_pair_p.append(int(content_p[1].split()[0]) / 2)

# Create a new DataFrame df_mapped
df_mapped = pd.DataFrame()
df_mapped["sample"] = SAMPLES
df_mapped["mapped_f"] = mapped_pair_f
df_mapped["mapped_u"] = mapped_pair_u
df_mapped["mapped_v"] = mapped_pair_v
df_mapped["mapped_a"] = mapped_pair_a
df_mapped["mapped_p"] = mapped_pair_p

# Merge df_mapped with reads_df based on the "sample" column
df_mapped = df_mapped.merge(reads_df, left_on="sample", right_on="sample")

# Add assembled bases from the QUAST statistics used in 03_assembly_short.tot.ipynb
assembly_stats_path = "/home/lmf/PhylloVir/VIRAL_WORLD/VIRAL_FRACTION/03_CONTIGS/statistics_quast_tot/transposed_report.tsv"
assembly_stats = pd.read_csv(assembly_stats_path, sep="\t")
assembly_stats["sample"] = (
    assembly_stats["Assembly"]
    .str.split("_spades").str[0]
)
assembly_stats = assembly_stats[["sample", "Total length (>= 0 bp)"]].rename(
    columns={"Total length (>= 0 bp)": "assembled_bases"}
)

df_mapped = df_mapped.merge(assembly_stats, on="sample", how="left")
df_mapped["expected coverage"] = df_mapped["mapped_a"] * 150 * 2 / df_mapped["assembled_bases"]
df_mapped["expected coverage"] = df_mapped["expected coverage"].replace([np.inf, -np.inf], np.nan)

# Calculate "% assembled", "% viral", "% unfiltered", and "% filtered" based on the merged DataFrame columns
df_mapped["% assembled"] = df_mapped["mapped_a"] * 100 / df_mapped["bbduk2"]
df_mapped["% viral"] = df_mapped["mapped_v"] * 100 / df_mapped["bbduk2"]
df_mapped["% unfiltered"] = df_mapped["mapped_u"] * 100 / df_mapped["bbduk2"]
df_mapped["% filtered"] = df_mapped["mapped_f"] * 100 / df_mapped["read_count"]
df_mapped["% active"] = df_mapped["mapped_p"] * 100 / df_mapped["read_count"]

# Calculate additional columns based on the merged DataFrame columns
df_mapped["filtered-viral"] = df_mapped["% filtered"] - df_mapped["% viral"]
df_mapped["filtered-unfiltered"] = df_mapped["% filtered"] - df_mapped["% unfiltered"]

# Calculate the mean row and append it to df_mapped
df_mapped.loc['mean'] = df_mapped.mean(numeric_only=True)
df_mapped.loc['min'] = df_mapped.min(numeric_only=True)
df_mapped.loc['max'] = df_mapped.max(numeric_only=True)
df_mapped.loc['std'] = df_mapped.std(numeric_only=True)

# Drop unnecessary columns from df_mapped
df_mapped = df_mapped.drop(["mapped_u", "mapped_v", "mapped_a", "bbduk2"], axis=1)

# Round values in df_mapped to 1 decimal place
df_mapped = df_mapped.round(1)

# Create a styled DataFrame stats_df_out with background color gradient
# stats_df_out = df_mapped.style.set_precision(2).background_gradient(cmap="RdYlGn").render()

# # Write the HTML table to a file
# with open(output_mapping_stats_html, "w") as fp:
#     fp.write(stats_df_out)

# # Display the styled DataFrame without writing to a file
df_mapped.round(2).style.background_gradient(cmap="RdYlGn")
```

```python
SAMPLES_mean = [
    'vFL', 'vOL', 'vF23', 
    'vN1', 'vN2', 'vN3', 
    'vZ1', 'vZ2', 'vZ3', 
    'vF1A', 'vF2A', 'vF3A', 
    'vF1B', 'vF2B', 'vF3B', 
    'vF1C', 'vF2C', 'vF3C', 
    'vO1A', 'vO2A', 'vO3A', 
    'vO1B', 'vO2B', 'vO3B', 
    'vO1C', 'vO2C', 'vO3C', 
    'vS1B', 'vS2B', 'vS3B', 
    'vS1C', 'vS2C', 'vS3C', 
    'vNWd', 'vSWd', 'vSEd',
]

df_mapped_viral_fraction=df_mapped[df_mapped["sample"].isin(SAMPLES_mean)].set_index("sample")
df_mapped_viral_fraction.loc['mean'] = df_mapped_viral_fraction.mean(numeric_only=True)
df_mapped_viral_fraction.loc['min'] = df_mapped_viral_fraction.min(numeric_only=True)
df_mapped_viral_fraction.loc['max'] = df_mapped_viral_fraction.max(numeric_only=True)
df_mapped_viral_fraction.loc['std'] = df_mapped_viral_fraction.std(numeric_only=True)

df_mapped_viral_fraction[["mapped_p", "mapped_f"]].to_csv("/home/lmf/PhylloVir/VIRAL_WORLD/VIRAL_FRACTION/FIGURES_AND_TABLES/mapped_reads_active.csv")
df_mapped_viral_fraction.round(2).style.background_gradient(cmap="RdYlGn")
```

```python
# Use all samples, or replace with the subset you want
# df_stats = df_mapped.copy()
df_stats = df_mapped[df_mapped["sample"].isin(SAMPLES_mean)].copy()

# Recalculate percentages to be explicit
df_stats["% filtered"] = df_stats["mapped_f"] * 100 / df_stats["read_count"]
df_stats["% active"] = df_stats["mapped_p"] * 100 / df_stats["read_count"]

# Summary stats for filtered and active
summary = df_stats[["% filtered", "% active"]].agg(["mean", "min", "max", "std"]).round(1)

# Ratio of means
ratio_of_means = (df_stats["% active"].mean() * 100 / df_stats["% filtered"].mean()).round(1)

summary
print(ratio_of_means, "%")

```

## Final Mapping Summary

The saved table contains the canonical 36 samples plus summary rows.

```python
# FINAL_MAPPING_PERCENTAGE_SUMMARY_BLOCK

VIRAL_FRACTION_DIR = (
    Path(mapping_dir).resolve().parent
    if "mapping_dir" in globals()
    else Path("/home/lmf/PhylloVir/VIRAL_WORLD/VIRAL_FRACTION")
)
MAPPING_DIR = (
    Path(mapping_dir).resolve()
    if "mapping_dir" in globals()
    else VIRAL_FRACTION_DIR / "06_MAPPING"
)
STATS_DIR = MAPPING_DIR / "STATS_FILES"
REFERENCES_DIR = MAPPING_DIR / "REFERENCES"
FIGURES_TABLES_DIR = VIRAL_FRACTION_DIR / "FIGURES_AND_TABLES"
SUP_FIGS_TABLES_DIR = FIGURES_TABLES_DIR / "SUP_FIGS_TABLES"
SUP_FIGS_TABLES_DIR.mkdir(parents=True, exist_ok=True)
SAMPLING = sampling if "sampling" in globals() else "tot"

QC_READ_COUNTS_PATH = (
    Path(qc_read_counts)
    if "qc_read_counts" in globals()
    else FIGURES_TABLES_DIR / f"01_qc_read_counts_paired.{SAMPLING}.csv"
)

FINAL_MAPPING_36_WITH_STATS_PATH = FIGURES_TABLES_DIR / "final_mapping_table_36_samples_with_stats.csv"

DEFAULT_FINAL_SAMPLES = [
    "vFL", "vOL", "vF23",
    "vN1", "vN2", "vN3",
    "vZ1", "vZ2", "vZ3",
    "vF1A", "vF2A", "vF3A",
    "vF1B", "vF2B", "vF3B",
    "vF1C", "vF2C", "vF3C",
    "vO1A", "vO2A", "vO3A",
    "vO1B", "vO2B", "vO3B",
    "vO1C", "vO2C", "vO3C",
    "vS1B", "vS2B", "vS3B",
    "vS1C", "vS2C", "vS3C",
    "vNWd", "vSWd", "vSEd",
]
FINAL_SAMPLES = list(SAMPLES_mean) if "SAMPLES_mean" in globals() else DEFAULT_FINAL_SAMPLES


def mapped_pairs_from_flagstats(path):
    """Estimate mapped read pairs from a samtools flagstats file."""
    path = Path(path)
    if not path.exists():
        return np.nan

    lines = path.read_text().splitlines()
    for line in lines:
        if " primary mapped" in line:
            return int(line.split()[0]) / 2

    # These filtered BAMs usually contain mapped reads only. Keep the primary
    # line as a fallback for short flagstats files.
    for line in lines:
        if " primary" in line and "duplicates" not in line:
            return int(line.split()[0]) / 2

    return np.nan


def pct(numerator, denominator):
    if pd.isna(numerator) or pd.isna(denominator) or denominator <= 0:
        return np.nan
    return numerator * 100 / denominator


def sample_from_assembly_flagstats(path):
    prefix = "bowtie2_flagstats_filtered_"
    suffix = f"_assembled_contigs.{SAMPLING}.txt"
    name = Path(path).name
    if name.startswith(prefix) and name.endswith(suffix):
        return name[len(prefix):-len(suffix)]
    return None


qc_reads = pd.read_csv(QC_READ_COUNTS_PATH, index_col=0)
read_counts = (
    qc_reads[["sample", "bbduk"]]
    .drop_duplicates("sample")
    .assign(bbduk=lambda df: pd.to_numeric(df["bbduk"], errors="coerce"))
    .set_index("sample")["bbduk"]
)

if "SAMPLES" in globals():
    ALL_SAMPLES = list(SAMPLES)
else:
    ALL_SAMPLES = sorted(
        sample
        for sample in (
            sample_from_assembly_flagstats(path)
            for path in STATS_DIR.glob(f"bowtie2_flagstats_filtered_*_assembled_contigs.{SAMPLING}.txt")
        )
        if sample is not None
    )

flagstats_templates = {
    "assembled": STATS_DIR / "bowtie2_flagstats_filtered_{sample}_assembled_contigs.{sampling}.txt",
    "viral": STATS_DIR / "bowtie2_flagstats_filtered_{sample}_viral_contigs.{sampling}.txt",
    "clustered": STATS_DIR / "bowtie2_flagstats_filtered_{sample}_unfiltered_contigs.{sampling}.txt",
    "filtered": MAPPING_DIR / "bowtie2_flagstats_filtered_{sample}.{sampling}.txt",
    "active": REFERENCES_DIR / "bowtie2_flagstats_filtered_active_vOTUs_threshold_1p5_{sample}.{sampling}.txt",
    "masked_strains": REFERENCES_DIR / "bowtie2_flagstats_filtered_strains_in_microbial_fraction_2018_geNomad_virus_masked_{sample}.{sampling}.txt",
}

percentage_cols = ["% assembled", "% viral", "% clustered", "% filtered", "% active", "% masked strains"]
report_cols = ["sample", "Read count"] + percentage_cols
numeric_cols = ["Read count"] + percentage_cols


def build_mapping_table(samples):
    rows = []
    missing_rows = []
    for sample in samples:
        read_count = read_counts.get(sample, np.nan)
        assembly_denominator = min(2_000_000, read_count) if pd.notna(read_count) else np.nan

        mapped = {}
        for label, template in flagstats_templates.items():
            path = Path(str(template).format(sample=sample, sampling=SAMPLING))
            mapped[label] = mapped_pairs_from_flagstats(path)
            if not path.exists():
                missing_rows.append({"sample": sample, "mapping_category": label, "path": str(path)})

        rows.append({
            "sample": sample,
            "Read count": read_count,
            "% assembled": pct(mapped["assembled"], assembly_denominator),
            "% viral": pct(mapped["viral"], assembly_denominator),
            "% clustered": pct(mapped["clustered"], assembly_denominator),
            "% filtered": pct(mapped["filtered"], read_count),
            "% active": pct(mapped["active"], read_count),
            "% masked strains": pct(mapped["masked_strains"], read_count),
        })
    return pd.DataFrame(rows, columns=report_cols), missing_rows


def format_mapping_table(table):
    table = table.copy()
    table[percentage_cols] = table[percentage_cols].round(1)
    table["Read count"] = pd.to_numeric(table["Read count"], errors="coerce").round(0).astype("Int64")
    return table


final_mapping_table_all_samples_raw, missing_all = build_mapping_table(ALL_SAMPLES)
final_mapping_table_36_samples_raw, missing_36 = build_mapping_table(FINAL_SAMPLES)

stats_rows = (
    final_mapping_table_36_samples_raw[numeric_cols]
    .agg(["mean", "min", "max", "std"])
    .reset_index()
    .rename(columns={"index": "sample"})
)
final_mapping_table_36_samples_with_stats_raw = pd.concat(
    [final_mapping_table_36_samples_raw, stats_rows],
    ignore_index=True,
)

final_mapping_table_all_samples = format_mapping_table(final_mapping_table_all_samples_raw)
final_mapping_table_36_samples_with_stats = format_mapping_table(final_mapping_table_36_samples_with_stats_raw)
missing_flagstats = (
    pd.DataFrame(missing_all + missing_36, columns=["sample", "mapping_category", "path"])
    .drop_duplicates()
)

FIGURES_TABLES_DIR.mkdir(parents=True, exist_ok=True)
final_mapping_table_36_samples_with_stats.to_csv(FINAL_MAPPING_36_WITH_STATS_PATH, index=False)
final_mapping_table_36_samples_with_stats.to_csv(
    SUP_FIGS_TABLES_DIR / "STable9_mapping_statistics_36_viromes.csv",
    index=False,
)

print(f"All-sample table rows displayed: {len(final_mapping_table_all_samples)}")
print(f"Saved 36-sample table with stats: {FINAL_MAPPING_36_WITH_STATS_PATH}")
print(f"Missing flagstats rows: {len(missing_flagstats)}")

# Keep the compact 36-sample summary as the interactive preview.
final_mapping_table_36_samples_with_stats
```

## Assembled Nonviral Bases by Species

Species-level assembled bases across the canonical 36 viromes.

```python
# NONVIRAL_SPECIES_ASSEMBLED_BP_MATRIX_BLOCK

VIRAL_FRACTION_DIR = (
    Path(mapping_dir).resolve().parent
    if "mapping_dir" in globals()
    else Path("/home/lmf/PhylloVir/VIRAL_WORLD/VIRAL_FRACTION")
)
ASSEMBLY_DIR = VIRAL_FRACTION_DIR / "03_CONTIGS"
FIGURES_TABLES_DIR = VIRAL_FRACTION_DIR / "FIGURES_AND_TABLES"
SUP_FIGS_TABLES_DIR = FIGURES_TABLES_DIR / "SUP_FIGS_TABLES"
SUP_FIGS_TABLES_DIR.mkdir(parents=True, exist_ok=True)

POSITIVE_VIRAL_CONTIG_LIST_PATH = (
    VIRAL_FRACTION_DIR / "04_VIRAL_ID" / "combined_positive_viral_contigs_list.csv"
)
MICROBIAL_TAXONOMY_PATH = (
    ASSEMBLY_DIR / "ALL_assembled_filtered_scaffolds_MICROBIAL_10000_taxonomy.tot.csv"
)
SOURMASH_REFERENCE_TAXONOMY_PATH = (
    Path("/home/lmf/db/SOURMASH_25_08/taxonomy/MATRIX_taxonomy.csv")
)
FINAL_MAPPING_36_WITH_STATS_PATH = (
    FIGURES_TABLES_DIR / "final_mapping_table_36_samples_with_stats.csv"
)
OUTPUT_PATH = (
    FIGURES_TABLES_DIR
    / "nonviral_species_assembled_bp_by_sample_36_viral_samples.csv"
)

MIN_CONTIG_LENGTH = 10_000
SUMMARY_ROW_LABELS = {"mean", "min", "max", "std"}
TAXONOMY_COLUMNS = ["phylum", "class", "order", "family", "genus", "species"]


def base_contig_id(value):
    return str(value).strip().lstrip(">").split()[0].split("_provirus")[0]


# Canonical biological viral cohort; remove the four summary rows from the saved table.
if "FINAL_SAMPLES" in globals():
    target_viral_samples = list(dict.fromkeys(map(str, FINAL_SAMPLES)))
else:
    final_mapping_36 = pd.read_csv(FINAL_MAPPING_36_WITH_STATS_PATH)
    target_viral_samples = [
        str(value)
        for value in final_mapping_36["sample"].dropna()
        if str(value).strip().lower() not in SUMMARY_ROW_LABELS
    ]
target_viral_samples = list(dict.fromkeys(target_viral_samples))
if len(target_viral_samples) != 36:
    raise ValueError(
        f"Expected 36 canonical viral samples, found {len(target_viral_samples)}"
    )
target_viral_sample_set = set(target_viral_samples)

# Confirm that the assembly exists for every target sample. Covstats are deliberately
# not used: the requested values are assembled sequence lengths, not mapping coverage.
missing_assemblies = [
    sample
    for sample in target_viral_samples
    if not (ASSEMBLY_DIR / f"{sample}_spades_filtered_scaffolds.tot.fasta").exists()
]
if missing_assemblies:
    raise FileNotFoundError(
        "Missing assemblies for target samples: " + ", ".join(missing_assemblies)
    )

# Exclude assembled contigs retained as positive viral sequences.
positive_viral_ids = pd.read_csv(POSITIVE_VIRAL_CONTIG_LIST_PATH, header=None)[0]
positive_viral_base_ids = set(positive_viral_ids.dropna().map(base_contig_id))

# Parse sample, contig length, and species directly from the microbial taxonomy table.
# Only successful species-resolved assignments from >=10 kb nonviral contigs contribute.
microbial_taxonomy = pd.read_csv(MICROBIAL_TAXONOMY_PATH)
tax_lineage = (
    microbial_taxonomy["lineage"]
    .fillna("")
    .str.split(";", expand=True)
    .reindex(columns=range(8))
    .replace({"": pd.NA, "N/A": pd.NA})
)
species_assignments = pd.DataFrame({
    "base_contig": microbial_taxonomy["query_name"].map(base_contig_id),
    "sample": microbial_taxonomy["query_name"].astype(str).str.extract(r"^(.+?)_NODE_")[0],
    "assembled_bp": pd.to_numeric(
        microbial_taxonomy["query_name"].astype(str).str.extract(r"_length_(\d+)_cov_")[0],
        errors="coerce",
    ),
    "tax_status": microbial_taxonomy["status"],
    "phylum": tax_lineage[1],
    "class": tax_lineage[2],
    "order": tax_lineage[3],
    "family": tax_lineage[4],
    "genus": tax_lineage[5],
    "species": tax_lineage[6],
}).drop_duplicates("base_contig")
species_assignments = species_assignments[
    species_assignments["sample"].isin(target_viral_sample_set)
    & ~species_assignments["base_contig"].isin(positive_viral_base_ids)
    & (species_assignments["assembled_bp"] >= MIN_CONTIG_LENGTH)
    & species_assignments["tax_status"].eq("match")
    & species_assignments["species"].notna()
]
species_assignments[TAXONOMY_COLUMNS] = (
    species_assignments[TAXONOMY_COLUMNS].fillna("Unresolved")
)

species_bp_long = (
    species_assignments
    .groupby(TAXONOMY_COLUMNS + ["sample"], observed=True)["assembled_bp"]
    .sum()
    .reset_index()
)
nonviral_species_assembled_bp_by_sample = (
    species_bp_long
    .pivot_table(
        index=TAXONOMY_COLUMNS,
        columns="sample",
        values="assembled_bp",
        aggfunc="sum",
        fill_value=0,
        observed=True,
    )
    .reindex(columns=target_viral_samples, fill_value=0)
    .fillna(0)
    .astype(int)
    .reset_index()
)

# Add the requested prevalence and total-base summaries, then order the rows.
sample_columns = target_viral_samples
nonviral_species_assembled_bp_by_sample["n_samples_with_assembled_bp"] = (
    (nonviral_species_assembled_bp_by_sample[sample_columns] > 0).sum(axis=1)
)
nonviral_species_assembled_bp_by_sample["total_assembled_bp"] = (
    nonviral_species_assembled_bp_by_sample[sample_columns].sum(axis=1)
)
nonviral_species_assembled_bp_by_sample = (
    nonviral_species_assembled_bp_by_sample[
        TAXONOMY_COLUMNS
        + ["n_samples_with_assembled_bp", "total_assembled_bp"]
        + sample_columns
    ]
    .sort_values(
        ["n_samples_with_assembled_bp", "total_assembled_bp"],
        ascending=[False, False],
    )
    .reset_index(drop=True)
)

# Add the species taxid to the publication copy. A single genome identifier
# is not added because this table is aggregated by species and may combine
# contigs assigned to many reference genomes or strains.
if not SOURMASH_REFERENCE_TAXONOMY_PATH.exists():
    raise FileNotFoundError(SOURMASH_REFERENCE_TAXONOMY_PATH)
species_taxid_reference = pd.read_csv(
    SOURMASH_REFERENCE_TAXONOMY_PATH,
    usecols=["taxid", "species"],
).dropna(subset=["taxid", "species"])
species_taxid_reference["species"] = (
    species_taxid_reference["species"].astype(str).str.strip()
)
species_taxid_reference = species_taxid_reference.loc[
    species_taxid_reference["species"].isin(
        nonviral_species_assembled_bp_by_sample["species"]
    )
]
taxids_per_species = species_taxid_reference.groupby("species")["taxid"].nunique()
ambiguous_species_taxids = taxids_per_species.loc[taxids_per_species > 1]
if not ambiguous_species_taxids.empty:
    raise ValueError(
        "Multiple taxids found for species in the nonviral publication table: "
        f"{ambiguous_species_taxids.index[:5].tolist()}"
    )
species_taxid_lookup = species_taxid_reference.drop_duplicates("species")[[
    "species", "taxid"
]]
nonviral_species_publication = nonviral_species_assembled_bp_by_sample.merge(
    species_taxid_lookup,
    on="species",
    how="left",
    validate="many_to_one",
)
if nonviral_species_publication["taxid"].isna().any():
    missing_taxid_species = nonviral_species_publication.loc[
        nonviral_species_publication["taxid"].isna(), "species"
    ].head().tolist()
    raise ValueError(
        f"Missing sourmash taxids for publication species: {missing_taxid_species}"
    )
nonviral_species_publication["taxid"] = (
    nonviral_species_publication["taxid"].astype(int)
)
nonviral_species_publication = nonviral_species_publication[
    ["taxid"]
    + TAXONOMY_COLUMNS
    + ["n_samples_with_assembled_bp", "total_assembled_bp"]
    + sample_columns
]

FIGURES_TABLES_DIR.mkdir(parents=True, exist_ok=True)
nonviral_species_assembled_bp_by_sample.to_csv(OUTPUT_PATH, index=False)
nonviral_species_publication.to_csv(
    SUP_FIGS_TABLES_DIR / "STable17_nonviral_contig_species_36_viromes.csv",
    index=False,
)

print(f"Species in matrix: {len(nonviral_species_assembled_bp_by_sample):,}")
print(f"Viral samples in matrix: {len(sample_columns)}")
print("vZ1 species-assigned nonviral assembled bp: " + str(
    int(nonviral_species_assembled_bp_by_sample["vZ1"].sum())
))
print(f"Saved: {OUTPUT_PATH}")
nonviral_species_assembled_bp_by_sample.head(50)
```

## Strain-Reference Genomes

```python
# STRAIN_REFERENCE_COVERAGE_BLOCK

VIRAL_FRACTION_DIR = (
    Path(mapping_dir).resolve().parent
    if "mapping_dir" in globals()
    else Path("/home/lmf/PhylloVir/VIRAL_WORLD/VIRAL_FRACTION")
)
VIRAL_WORLD_DIR = VIRAL_FRACTION_DIR.parent
BACTERIAL_FRACTION_DIR = VIRAL_WORLD_DIR / "BACTERIAL_FRACTION"
MAPPING_DIR = (
    Path(mapping_dir).resolve()
    if "mapping_dir" in globals()
    else VIRAL_FRACTION_DIR / "06_MAPPING"
)
REFERENCES_DIR = MAPPING_DIR / "REFERENCES"
FIGURES_TABLES_DIR = VIRAL_FRACTION_DIR / "FIGURES_AND_TABLES"
SAMPLING = sampling if "sampling" in globals() else "tot"

QC_READ_COUNTS_PATH = (
    Path(qc_read_counts)
    if "qc_read_counts" in globals()
    else FIGURES_TABLES_DIR / "01_qc_read_counts_paired.tot.csv"
)
STRAIN_CONTIG_METADATA_PATH = (
    BACTERIAL_FRACTION_DIR
    / "MICROBIAL_GENOMES_PHYLLOVIR"
    / "strains_in_microbial_fraction_2018.seqtk_comp.with_source.tsv"
)
STRAIN_METADATA_PATH = (
    BACTERIAL_FRACTION_DIR
    / "06_MAPPING"
    / "strains_in_microbial_fraction_metadata.csv"
)

# User threshold: a strain/genome must have this much of its reference length covered.
# Canonical high-confidence detection threshold used by Supplementary Notebook 08.
# This is a fraction, not a percent: 0.30 means 30% genome breadth covered.
MIN_STRAIN_GENOME_COVERED_FRACTION = 0.30

# Optional read-count floor after genome-level aggregation.
# Keep at 1 to report every strain/genome passing the coverage threshold.
MIN_STRAIN_READ_COUNT = 1


def sample_from_strain_covstats_path(path):
    pattern = rf"^bowtie2_strains_in_microbial_fraction_2018_geNomad_virus_masked_(.+)_{SAMPLING}_covstats\.txt$"
    match = re.match(pattern, path.name)
    return match.group(1) if match else None


def read_strain_covstats(path):
    df = pd.read_csv(path, sep="\t")
    if df.shape[1] != 8:
        raise ValueError(f"Expected 8 covstats columns in {path}, found {df.shape[1]}")
    df.columns = [
        "contig", "mean", "length", "covered_bases", "read_count",
        "variance", "trimmed_mean", "rpkm",
    ]
    numeric_cols = ["mean", "length", "covered_bases", "read_count", "variance", "trimmed_mean", "rpkm"]
    df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors="coerce")
    df["contig"] = df["contig"].astype(str).str.strip()
    return df


# Map strain-reference contig IDs back to their parent genome/strain IDs.
strain_contig_cols = [
    "seq", "seq_length", "A", "C", "G", "T", "#2", "#3", "#4",
    "#CpG", "#tv", "#ts", "#CpG-ts", "genome",
]
strain_contig_metadata = pd.read_csv(
    STRAIN_CONTIG_METADATA_PATH,
    sep="\t",
    names=strain_contig_cols,
    usecols=["seq", "genome"],
)
seq_to_genome = strain_contig_metadata.drop_duplicates("seq").set_index("seq")["genome"]

# Sample read denominator. `bbduk` is the R1/pair count in this paired-read QC table;
# covstats Read Count is reads, so multiply by 2 to estimate total cleaned reads.
if QC_READ_COUNTS_PATH.exists():
    qc_reads = pd.read_csv(QC_READ_COUNTS_PATH, index_col=0)
    clean_read_denominators = (
        qc_reads[["sample", "bbduk"]]
        .drop_duplicates("sample")
        .assign(clean_reads=lambda df: pd.to_numeric(df["bbduk"], errors="coerce") * 2)
        .set_index("sample")["clean_reads"]
    )
else:
    clean_read_denominators = pd.Series(dtype=float)

strain_metadata_cols = [
    "ident", "taxid", "superkingdom", "kingdom", "phylum", "class",
    "order", "family", "genus", "species", "strain", "genome_size",
]
strain_metadata = pd.read_csv(STRAIN_METADATA_PATH)
strain_metadata_cols = [col for col in strain_metadata_cols if col in strain_metadata.columns]
strain_metadata = strain_metadata[strain_metadata_cols].drop_duplicates("ident")

strain_coverage_rows = []
for covstats_path in sorted(REFERENCES_DIR.glob(f"bowtie2_strains_in_microbial_fraction_2018_geNomad_virus_masked*_{SAMPLING}_covstats.txt")):
    sample = sample_from_strain_covstats_path(covstats_path)
    if sample is None:
        continue

    covstats = read_strain_covstats(covstats_path)
    covstats["genome"] = covstats["contig"].map(seq_to_genome).fillna(covstats["contig"])
    covstats["has_mapped_reads"] = covstats["read_count"] > 0
    total_strain_reference_reads = covstats["read_count"].sum()
    clean_reads = clean_read_denominators.get(sample, np.nan)

    genome_coverage = (
        covstats
        .groupby("genome", as_index=False)
        .agg(
            genome_length=("length", "sum"),
            covered_bases=("covered_bases", "sum"),
            read_count=("read_count", "sum"),
            n_reference_contigs=("contig", "nunique"),
            n_contigs_with_reads=("has_mapped_reads", "sum"),
            max_rpkm=("rpkm", "max"),
            median_rpkm=("rpkm", "median"),
            max_trimmed_mean=("trimmed_mean", "max"),
            median_trimmed_mean=("trimmed_mean", "median"),
        )
    )
    genome_coverage["sample"] = sample
    genome_coverage["genome_covered_fraction"] = (
        genome_coverage["covered_bases"] / genome_coverage["genome_length"]
    )
    genome_coverage["genome_covered_percentage"] = genome_coverage["genome_covered_fraction"] * 100
    genome_coverage["pct_of_clean_reads"] = np.where(
        clean_reads > 0,
        genome_coverage["read_count"] * 100 / clean_reads,
        np.nan,
    )
    genome_coverage["pct_of_strain_reference_mapped_reads"] = np.where(
        total_strain_reference_reads > 0,
        genome_coverage["read_count"] * 100 / total_strain_reference_reads,
        np.nan,
    )
    genome_coverage["total_strain_reference_read_count"] = total_strain_reference_reads
    strain_coverage_rows.append(genome_coverage)

if strain_coverage_rows:
    strain_reference_coverage = pd.concat(strain_coverage_rows, ignore_index=True)
    strain_reference_coverage = strain_reference_coverage.merge(
        strain_metadata,
        left_on="genome",
        right_on="ident",
        how="left",
    )
else:
    strain_reference_coverage = pd.DataFrame()

if not strain_reference_coverage.empty:
    strain_reference_breadth30 = (
        strain_reference_coverage[
            (strain_reference_coverage["genome_covered_fraction"] >= MIN_STRAIN_GENOME_COVERED_FRACTION)
            & (strain_reference_coverage["read_count"] >= MIN_STRAIN_READ_COUNT)
        ]
        .sort_values(["pct_of_clean_reads", "genome_covered_percentage", "read_count"], ascending=False)
        .copy()
    )
else:
    strain_reference_breadth30 = pd.DataFrame()

if not strain_reference_breadth30.empty:
    strain_reference_breadth30_sample_summary = (
        strain_reference_breadth30
        .groupby("sample", as_index=False)
        .agg(
            n_strains_breadth30=("genome", "nunique"),
            total_read_count=("read_count", "sum"),
            total_pct_of_clean_reads=("pct_of_clean_reads", "sum"),
            max_genome_covered_percentage=("genome_covered_percentage", "max"),
            top_genus=("genus", lambda values: values.dropna().iloc[0] if len(values.dropna()) else np.nan),
            top_strain=("strain", lambda values: values.dropna().iloc[0] if len(values.dropna()) else np.nan),
        )
        .sort_values("total_pct_of_clean_reads", ascending=False)
    )

    strain_reference_breadth30_taxonomy_summary = (
        strain_reference_breadth30
        .groupby(["phylum", "class", "order", "family", "genus", "species"], dropna=False)
        .agg(
            n_genomes=("genome", "nunique"),
            n_samples=("sample", "nunique"),
            samples=("sample", lambda values: ", ".join(sorted(set(map(str, values))))),
            total_read_count=("read_count", "sum"),
            total_pct_of_clean_reads=("pct_of_clean_reads", "sum"),
            median_genome_covered_percentage=("genome_covered_percentage", "median"),
            max_genome_covered_percentage=("genome_covered_percentage", "max"),
        )
        .reset_index()
        .sort_values("total_pct_of_clean_reads", ascending=False)
    )
else:
    strain_reference_breadth30_sample_summary = pd.DataFrame()
    strain_reference_breadth30_taxonomy_summary = pd.DataFrame()

FIGURES_TABLES_DIR.mkdir(parents=True, exist_ok=True)
strain_reference_breadth30.to_csv(
    FIGURES_TABLES_DIR / "strain_reference_genomes_breadth30.csv",
    index=False,
)
strain_reference_breadth30_sample_summary.to_csv(
    FIGURES_TABLES_DIR / "strain_reference_breadth30_sample_summary.csv",
    index=False,
)
strain_reference_breadth30_taxonomy_summary.to_csv(
    FIGURES_TABLES_DIR / "strain_reference_breadth30_taxonomy_summary.csv",
    index=False,
)

print(
    "Strain/genome hits with >="
    f"{MIN_STRAIN_GENOME_COVERED_FRACTION:.0%} breadth covered: "
    f"{len(strain_reference_breadth30)} rows from "
    f"{strain_reference_breadth30['sample'].nunique() if not strain_reference_breadth30.empty else 0} samples"
)

sample_display_cols = [
    "sample", "n_strains_breadth30", "total_read_count", "total_pct_of_clean_reads",
    "max_genome_covered_percentage", "top_genus", "top_strain",
]
sample_display_cols = [col for col in sample_display_cols if col in strain_reference_breadth30_sample_summary.columns]
strain_reference_breadth30_sample_summary[sample_display_cols].head(30)

strain_display_cols = [
    "sample", "genome", "strain", "genome_length", "covered_bases", "genome_covered_percentage",
    "read_count", "pct_of_clean_reads", "pct_of_strain_reference_mapped_reads",
    "n_reference_contigs", "n_contigs_with_reads", "phylum", "family", "genus", "species",
]
strain_display_cols = [col for col in strain_display_cols if col in strain_reference_breadth30.columns]
strain_reference_breadth30[strain_display_cols].head(100)

taxonomy_display_cols = [
    "phylum", "family", "genus", "species", "n_genomes", "n_samples", "samples",
    "total_read_count", "total_pct_of_clean_reads", "median_genome_covered_percentage",
]
taxonomy_display_cols = [col for col in taxonomy_display_cols if col in strain_reference_breadth30_taxonomy_summary.columns]
strain_reference_breadth30_taxonomy_summary[taxonomy_display_cols].head(50)
```

## Coverage-Ranked Strain Heatmap

Colors show selected-genome RPKM; annotations show genome breadth (%).

```python
# TOP50_MAPPING_STRAINS_RPKM_HEATMAP_BLOCK

VIRAL_FRACTION_DIR = (
    Path(mapping_dir).resolve().parent
    if "mapping_dir" in globals()
    else Path("/home/lmf/PhylloVir/VIRAL_WORLD/VIRAL_FRACTION")
)
FIGURES_TABLES_DIR = VIRAL_FRACTION_DIR / "FIGURES_AND_TABLES"

STRAIN_REFERENCE_COVERAGE_PATH = FIGURES_TABLES_DIR / "strain_reference_genomes_breadth30.csv"
COVERAGE_RANKED_RPKM_MATRIX_PATH = FIGURES_TABLES_DIR / "top_species_coverage_ranked_rpkm_matrix.csv"
COVERAGE_RANKED_COVERAGE_MATRIX_PATH = FIGURES_TABLES_DIR / "top_species_coverage_ranked_genome_coverage_matrix.csv"
COVERAGE_RANKED_SELECTED_GENOME_MATRIX_PATH = FIGURES_TABLES_DIR / "top_species_coverage_ranked_selected_genome_matrix.csv"
COVERAGE_RANKED_SELECTED_STRAIN_MATRIX_PATH = FIGURES_TABLES_DIR / "top_species_coverage_ranked_selected_strain_matrix.csv"
COVERAGE_RANKED_ROW_METADATA_PATH = FIGURES_TABLES_DIR / "top_species_coverage_ranked_row_metadata.csv"
COVERAGE_RANKED_HEATMAP_PNG = FIGURES_TABLES_DIR / "top_species_coverage_ranked_rpkm_heatmap.png"
COVERAGE_RANKED_HEATMAP_SVG = FIGURES_TABLES_DIR / "top_species_coverage_ranked_rpkm_heatmap.svg"

# User threshold: number of species rows to show in the heatmap.
TOP_N_SPECIES = 20

# User choice: select and rank species by genome breadth coverage.
COVERAGE_COLUMN = "genome_covered_percentage"

# User choice: keep RPKM as the heatmap color for the selected genome.
RPKM_COLOR_COLUMN = "max_rpkm"

# User choice: plot log10(RPKM + 1) so a few very abundant strains do not flatten the heatmap.
PLOT_LOG10_RPKM = True

# User display settings. Tick locations are forced below so labels are not skipped
# when the figure is resized smaller.
HEATMAP_XTICK_FONT_SIZE = 5
HEATMAP_YTICK_FONT_SIZE = 5
HEATMAP_AXIS_LABEL_FONT_SIZE = 7
HEATMAP_TITLE_FONT_SIZE = 7
HEATMAP_ANNOT_FONT_SIZE = 4

DEFAULT_HEATMAP_SAMPLES = [
    "vFL", "vOL", "vF23",
    "vN1", "vN2", "vN3",
    "vZ1", "vZ2", "vZ3",
    "vF1A", "vF2A", "vF3A",
    "vF1B", "vF2B", "vF3B",
    "vF1C", "vF2C", "vF3C",
    "vO1A", "vO2A", "vO3A",
    "vO1B", "vO2B", "vO3B",
    "vO1C", "vO2C", "vO3C",
    "vS1B", "vS2B", "vS3B",
    "vS1C", "vS2C", "vS3C",
    "vNWd", "vSWd", "vSEd",
]
HEATMAP_SAMPLES = (
    list(FINAL_SAMPLES)
    if "FINAL_SAMPLES" in globals()
    else list(SAMPLES_mean) if "SAMPLES_mean" in globals()
    else DEFAULT_HEATMAP_SAMPLES
)


def first_nonempty(values):
    values = pd.Series(values).dropna().astype(str)
    values = values[values.str.strip().ne("") & values.str.lower().ne("nan")]
    return values.iloc[0] if len(values) else np.nan


def clean_group(values, fallback):
    group = values.astype("string")
    group = group.mask(group.str.strip().eq(""), pd.NA)
    group = group.mask(group.str.lower().eq("nan"), pd.NA)
    return group.fillna(fallback.astype(str))


def add_taxonomic_groups(df):
    species_group = (
        clean_group(df["species"], df["genome"])
        if "species" in df.columns
        else df["genome"].astype(str)
    )
    if "genus" in df.columns:
        fallback_genus = species_group.astype(str).str.split().str[0]
        genus_group = clean_group(df["genus"], fallback_genus)
    else:
        genus_group = species_group.astype(str).str.split().str[0]
    df["species_group"] = species_group.astype(str)
    df["genus_group"] = genus_group.astype(str)
    return df


def species_matrix_out(matrix, row_metadata=None):
    out = matrix.reset_index().rename(columns={"species_group": "species"})
    if row_metadata is not None:
        metadata_cols = [
            "species", "row_label", "genus", "genus_total_coverage", "species_total_coverage",
            "species_max_coverage", "species_mean_coverage_detected",
            "top_coverage_genome", "top_coverage_strain", "top_strain_max_coverage",
            "top_strain_total_coverage", "n_samples_detected",
            "species_total_max_rpkm", "species_total_mapped_reads",
        ]
        metadata_cols = [col for col in metadata_cols if col in row_metadata.columns]
        out = out.merge(row_metadata[metadata_cols], on="species", how="left")
        sample_cols = [col for col in out.columns if col in HEATMAP_SAMPLES]
        out = out[[col for col in metadata_cols if col in out.columns] + sample_cols]
    return out


def annotation_strings(coverage_matrix, rpkm_matrix):
    coverage_for_labels = coverage_matrix.where(rpkm_matrix > 0)
    return coverage_for_labels.apply(
        lambda col: col.map(lambda value: "" if pd.isna(value) else f"{value:.0f}")
    )


if "strain_reference_breadth30" in globals() and not strain_reference_breadth30.empty:
    strain_source = strain_reference_breadth30.copy()
elif STRAIN_REFERENCE_COVERAGE_PATH.exists():
    strain_source = pd.read_csv(STRAIN_REFERENCE_COVERAGE_PATH)
else:
    strain_source = pd.DataFrame()

required_cols = {"sample", "genome", RPKM_COLOR_COLUMN, COVERAGE_COLUMN, "read_count"}
if strain_source.empty or not required_cols.issubset(strain_source.columns):
    print("No strain-reference RPKM/read-count/coverage table available for the heatmap.")
else:
    strain_source = strain_source.copy()
    strain_source[RPKM_COLOR_COLUMN] = pd.to_numeric(
        strain_source[RPKM_COLOR_COLUMN],
        errors="coerce",
    ).fillna(0)
    strain_source[COVERAGE_COLUMN] = pd.to_numeric(
        strain_source[COVERAGE_COLUMN],
        errors="coerce",
    ).fillna(0)
    strain_source["read_count"] = pd.to_numeric(
        strain_source["read_count"],
        errors="coerce",
    ).fillna(0)
    strain_source = strain_source[strain_source["sample"].isin(HEATMAP_SAMPLES)]
    strain_source = add_taxonomic_groups(strain_source)
    if "strain" not in strain_source.columns:
        strain_source["strain"] = np.nan

    # For each species and sample, keep the strain/reference genome with the
    # highest genome breadth coverage. Ties prefer higher RPKM, then more reads.
    species_sample_max = (
        strain_source
        .sort_values(
            ["species_group", "sample", COVERAGE_COLUMN, RPKM_COLOR_COLUMN, "read_count", "genome"],
            ascending=[True, True, False, False, False, True],
        )
        .drop_duplicates(["species_group", "sample"], keep="first")
    )

    species_signal = (
        species_sample_max
        .groupby(["genus_group", "species_group"], as_index=False)
        .agg(
            species_total_coverage=(COVERAGE_COLUMN, "sum"),
            species_max_coverage=(COVERAGE_COLUMN, "max"),
            species_mean_coverage_detected=(COVERAGE_COLUMN, "mean"),
            species_total_max_rpkm=(RPKM_COLOR_COLUMN, "sum"),
            species_total_mapped_reads=("read_count", "sum"),
            n_samples_detected=("sample", "nunique"),
        )
    )

    top_strain_by_coverage = (
        strain_source
        .groupby(["genus_group", "species_group", "genome"], as_index=False)
        .agg(
            top_strain_total_coverage=(COVERAGE_COLUMN, "sum"),
            top_strain_max_coverage=(COVERAGE_COLUMN, "max"),
            top_strain_total_max_rpkm=(RPKM_COLOR_COLUMN, "sum"),
            top_strain_mapped_reads=("read_count", "sum"),
            top_coverage_strain=("strain", first_nonempty),
        )
        .sort_values(
            ["species_group", "top_strain_total_coverage", "top_strain_max_coverage", "top_strain_total_max_rpkm", "top_strain_mapped_reads", "genome"],
            ascending=[True, False, False, False, False, True],
        )
        .drop_duplicates("species_group", keep="first")
        .rename(columns={"genome": "top_coverage_genome"})
    )

    row_metadata = (
        species_signal
        .merge(
            top_strain_by_coverage[
                [
                    "species_group", "top_coverage_genome", "top_coverage_strain",
                    "top_strain_total_coverage", "top_strain_max_coverage",
                    "top_strain_total_max_rpkm", "top_strain_mapped_reads",
                ]
            ],
            on="species_group",
            how="left",
        )
        .rename(columns={"species_group": "species", "genus_group": "genus"})
        .sort_values(
            ["species_total_coverage", "species_max_coverage", "n_samples_detected", "species_total_max_rpkm"],
            ascending=False,
        )
        .head(TOP_N_SPECIES)
    )

    genus_signal = (
        row_metadata
        .groupby("genus", as_index=False)
        .agg(
            genus_total_coverage=("species_total_coverage", "sum"),
            genus_max_coverage=("species_max_coverage", "max"),
            n_species_in_heatmap=("species", "nunique"),
        )
    )
    row_metadata = row_metadata.merge(genus_signal, on="genus", how="left")
    has_strain_label = (
        row_metadata["top_coverage_strain"].notna()
        & row_metadata["top_coverage_strain"].astype(str).str.strip().ne("")
        & row_metadata["top_coverage_strain"].astype(str).str.lower().ne("nan")
    )
    row_metadata["row_label"] = np.where(
        has_strain_label,
        row_metadata["top_coverage_strain"].astype(str),
        row_metadata["species"].astype(str),
    )

    # Final display order: genera by total coverage, then species by coverage.
    row_metadata = (
        row_metadata
        .assign(
            genus_sort=lambda df: df["genus"].str.lower(),
            species_sort=lambda df: df["species"].str.lower(),
        )
        .sort_values(
            ["genus_total_coverage", "genus_sort", "species_total_coverage", "species_sort"],
            ascending=[False, True, False, True],
        )
        .drop(columns=["genus_sort", "species_sort"])
        .reset_index(drop=True)
    )
    ordered_species = row_metadata["species"].tolist()

    species_sample_top = species_sample_max[species_sample_max["species_group"].isin(ordered_species)].copy()
    species_rpkm_matrix = (
        species_sample_top
        .pivot_table(
            index="species_group",
            columns="sample",
            values=RPKM_COLOR_COLUMN,
            aggfunc="max",
            fill_value=0,
        )
        .reindex(index=ordered_species)
        .reindex(columns=HEATMAP_SAMPLES, fill_value=0)
    )
    species_coverage_matrix = (
        species_sample_top
        .pivot_table(
            index="species_group",
            columns="sample",
            values=COVERAGE_COLUMN,
            aggfunc="max",
        )
        .reindex(index=ordered_species)
        .reindex(columns=HEATMAP_SAMPLES)
    )
    selected_genome_matrix = (
        species_sample_top
        .pivot_table(
            index="species_group",
            columns="sample",
            values="genome",
            aggfunc=first_nonempty,
        )
        .reindex(index=ordered_species)
        .reindex(columns=HEATMAP_SAMPLES)
    )
    selected_strain_matrix = (
        species_sample_top
        .pivot_table(
            index="species_group",
            columns="sample",
            values="strain",
            aggfunc=first_nonempty,
        )
        .reindex(index=ordered_species)
        .reindex(columns=HEATMAP_SAMPLES)
    )

    FIGURES_TABLES_DIR.mkdir(parents=True, exist_ok=True)
    row_metadata.to_csv(COVERAGE_RANKED_ROW_METADATA_PATH, index=False)
    species_matrix_out(species_rpkm_matrix, row_metadata).to_csv(COVERAGE_RANKED_RPKM_MATRIX_PATH, index=False)
    species_matrix_out(species_coverage_matrix.round(0).astype("Int64"), row_metadata).to_csv(
        COVERAGE_RANKED_COVERAGE_MATRIX_PATH,
        index=False,
    )
    species_matrix_out(selected_genome_matrix).to_csv(COVERAGE_RANKED_SELECTED_GENOME_MATRIX_PATH, index=False)
    species_matrix_out(selected_strain_matrix).to_csv(COVERAGE_RANKED_SELECTED_STRAIN_MATRIX_PATH, index=False)

    row_label_map = row_metadata.set_index("species")["row_label"].to_dict()
    plot_rpkm_matrix = species_rpkm_matrix.rename(index=row_label_map)
    plot_coverage_matrix = species_coverage_matrix.rename(index=row_label_map)
    heatmap_values = np.log10(plot_rpkm_matrix + 1) if PLOT_LOG10_RPKM else plot_rpkm_matrix
    colorbar_label = f"log10({RPKM_COLOR_COLUMN} + 1)" if PLOT_LOG10_RPKM else RPKM_COLOR_COLUMN
    coverage_annotation = annotation_strings(plot_coverage_matrix, plot_rpkm_matrix)

    fig_height = max(8, 0.28 * len(plot_rpkm_matrix))
    fig_width = max(14, 0.45 * len(plot_rpkm_matrix.columns))
    fig, ax = plt.subplots(figsize=(fig_width, fig_height))
    sns.heatmap(
        heatmap_values,
        cmap="viridis",
        linewidths=0,
        ax=ax,
        annot=coverage_annotation,
        fmt="",
        annot_kws={"fontsize": HEATMAP_ANNOT_FONT_SIZE},
        cbar_kws={"label": colorbar_label},
    )

    # Force every sample/species label to remain visible after resizing.
    ax.set_xticks(np.arange(len(plot_rpkm_matrix.columns)) + 0.5)
    ax.set_xticklabels(
        plot_rpkm_matrix.columns,
        rotation=90,
        ha="center",
        fontsize=HEATMAP_XTICK_FONT_SIZE,
    )
    ax.set_yticks(np.arange(len(plot_rpkm_matrix.index)) + 0.5)
    ax.set_yticklabels(
        plot_rpkm_matrix.index,
        rotation=0,
        fontsize=HEATMAP_YTICK_FONT_SIZE,
    )

    genus_order = row_metadata.set_index("species").loc[species_rpkm_matrix.index, "genus"]
    genus_boundaries = np.where(genus_order.ne(genus_order.shift()).to_numpy()[1:])[0] + 1
    for boundary in genus_boundaries:
        ax.hlines(boundary, *ax.get_xlim(), colors="white", linewidth=1)

    max_heat = np.nanmax(heatmap_values.to_numpy()) if heatmap_values.size else 0
    for text, value in zip(ax.texts, heatmap_values.to_numpy().ravel()):
        if not text.get_text():
            continue
        text.set_color("black" if max_heat and value > max_heat * 0.55 else "white")

    ax.set_xlabel("Sample", fontsize=HEATMAP_AXIS_LABEL_FONT_SIZE)
    ax.set_ylabel("Top coverage-ranked strain, grouped by genus", fontsize=HEATMAP_AXIS_LABEL_FONT_SIZE)
    ax.figure.axes[-1].yaxis.label.set_size(HEATMAP_AXIS_LABEL_FONT_SIZE)
    ax.figure.axes[-1].tick_params(labelsize=HEATMAP_XTICK_FONT_SIZE)
    plt.tight_layout()
    fig.savefig(COVERAGE_RANKED_HEATMAP_PNG, dpi=300, bbox_inches="tight")
    fig.savefig(COVERAGE_RANKED_HEATMAP_SVG, bbox_inches="tight")
    plt.show()

    print(f"Saved coverage-ranked row metadata: {COVERAGE_RANKED_ROW_METADATA_PATH}")
    print(f"Saved coverage-ranked RPKM matrix: {COVERAGE_RANKED_RPKM_MATRIX_PATH}")
    print(f"Saved coverage-ranked genome coverage matrix: {COVERAGE_RANKED_COVERAGE_MATRIX_PATH}")
    print(f"Saved selected-genome matrix: {COVERAGE_RANKED_SELECTED_GENOME_MATRIX_PATH}")
    print(f"Saved selected-strain matrix: {COVERAGE_RANKED_SELECTED_STRAIN_MATRIX_PATH}")
    print(f"Saved heatmap PNG: {COVERAGE_RANKED_HEATMAP_PNG}")
    print(f"Saved heatmap SVG: {COVERAGE_RANKED_HEATMAP_SVG}")
```
